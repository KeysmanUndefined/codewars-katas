try: ejecuta el código.

except: solo se ejecuta si se produce la excepción indicada.

finally: siempre se ejecuta.

else: no aparece en este ejemplo, por lo que no hay ningún bloque else que ejecutar.

'*': repite una cadena el número de veces indicado.

\n: introduce un salto de línea.

sep: almacena el separador para reutilizarlo en otros mensajes.


1. **sep:** crea una cadena formada por 45 guiones (`-`) y un salto de línea (`\n`).
2. **print():** muestra el contenido de `sep` seguido del texto `EXCEPTION RAISED AND CAUGHT`.

## Resultado esperado

EXCEPTION RAISED AND CAUGHT

```python
sep = '-' * 45 + '\n'
print(sep + 'EXCEPTION RAISED AND CAUGHT')
```

1. **try:** intenta ejecutar `x = 'spam'[99]`.
2. **IndexError:** Python genera una excepción porque la posición 99 está fuera de los límites de la cadena.
3. **except IndexError:** captura la excepción y ejecuta `print('except run')`.
4. **finally:** se ejecuta siempre, independientemente de que se produzca una excepción o no.
5. **print('after run'):** se ejecuta porque la excepción ha sido capturada y el programa puede continuar normalmente.

## Resultado esperado

except run
finally run
after run

```python
try:
    x = 'spam'[99]
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```

1. **print():** muestra el mensaje inicial.
2. **try:** intenta ejecutar `x = 'spam'[3]`.
3. **Resultado:** Python accede a la posición 3 de la cadena `'spam'` y obtiene el carácter `'m'`. No se produce ninguna excepción.
4. **except IndexError:** no se ejecuta porque no se ha producido ninguna excepción.
5. **finally:** se ejecuta siempre, aunque no se haya producido ningún error.
6. **print('after run'):** se ejecuta porque no hay ninguna excepción pendiente y el programa continúa normalmente.

1. **print():** muestra el mensaje inicial.
2. **try:** intenta ejecutar `x = 'spam'[3]`.
3. **Resultado:** Python accede a la posición 3 de la cadena `'spam'` y obtiene el carácter `'m'`. No se produce ninguna excepción.
4. **except IndexError:** no se ejecuta porque no se ha producido ningún error.
5. **else:** se ejecuta porque el bloque `try` ha terminado correctamente, sin excepciones.
6. **finally:** se ejecuta siempre, independientemente de que se haya producido una excepción o no.
7. **print('after run'):** se ejecuta porque no se ha producido ninguna excepción y el programa continúa normalmente.

NO EXCEPTION RAISED, WITH ELSE
else run

```python
print(sep + 'NO EXCEPTION RAISED, WITH ELSE')
try:
    x = 'spam'[3]
except IndexError:
    print('except run')
else:
    print('else run')
finally:
    print('finally run')
print('after run')
```

1. **print():** muestra el mensaje inicial.
2. **try:** intenta ejecutar `x = 1 / 0`.
3. **ZeroDivisionError:** Python genera una excepción porque no se puede dividir entre cero.
4. **except IndexError:** no se ejecuta porque el error no es de tipo `IndexError`.
5. **finally:** sí se ejecuta porque siempre se ejecuta al salir del bloque `try`.
6. **print('after run'):** no se ejecuta porque la excepción sigue sin capturarse y Python detiene el programa.

```python
print(sep + 'EXCEPTION RAISED BUT NOT CAUGHT')
try:
    x = 1 / 0
except IndexError:
    print('except run')
finally:
    print('finally run')
print('after run')
```
