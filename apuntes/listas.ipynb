# Listas

## 1. ¿Qué es una lista?
Una lista es una colección ordenada de elementos que pueden ser de diferentes tipos.

```python
mi_lista = [10, "hola", 3.14, True]
```

Una lista puede contener:

Números: 10, 25, 3.14

Cadenas de texto: "hola", "Python"

Booleanos: True, False

Otras listas: [1, 2, 3]

**Importante**: las listas pueden contener distintos tipos de datos al mismo tiempo.



## 2. Características principales de las listas

## 2.1 Ordered — Ordenadas
Los elementos mantienen un orden y cada uno tiene una posición.

```python
numeros = [10, 20, 30]
print(numeros)
```

## 2.2 Accessed by offset — Acceso mediante índices
Puedes acceder a un elemento utilizando su índice. En Python, los índices empiezan por 0.

```python
numeros = [10, 20, 30]

print(numeros[2])  
print(numeros[0]) 
print(numeros[1])  
```

También puedes usar índices negativos:

```python
numeros = [10, 20, 30]

print(numeros[-1])
```

## 2.3 Variable-length — Longitud variable
Las listas pueden crecer o reducirse. Puedes añadir y eliminar elementos.

```python
frutas = ["manzana", "pera"]

frutas.append("plátano")
print(frutas)
```

Para eliminar un elemento por su valor:

```python
frutas = ["manzana", "pera"]

frutas.remove("pera")
print(frutas)
```

## 2.4 Heterogeneous — Heterogéneas
Una lista puede contener diferentes tipos de datos.

```python
datos = [25, "Ana", 1.75, True]
```

Aquí tenemos un entero, una cadena, un número decimal y un booleano.

## 2.5 Arbitrarily nestable — Anidamiento
Una lista puede contener otras listas. 
Imprime la posicion 2 de la lista y la posicion 1 dentro de esta:

```python
matriz = [[1, 2], [2, 4]]
print(matriz[1][0])
```

## 3 ¿Qué significa que una lista sea mutable?

Una lista es mutable porque puedes modificar sus elementos después de crearla, sin tener que crear una lista nueva.

```python
numeros = [10, 20, 30]

numeros[0] = 99

print(numeros)
```

En cambio, las **cadenas de texto (str) son inmutables**: no puedes modificar directamente uno de sus caracteres.

## 5. ¿Qué son las referencias a objetos?
El libro explica que las listas almacenan referencias a objetos, no copias automáticas de esos objetos.

```python
a = [1, 2, 3]
b = a

b[0] = 99

print(a)
```

¿Por qué cambia también a? Porque a y b hacen referencia a la misma lista.

Si quieres crear una copia independiente de una lista sencilla, puedes utilizar:

```python
a = [1, 2, 3]
b = a.copy()

b[0] = 99

print(a) 
print(b)  
```

Qué debes **recordar**: asignar una lista a otra variable no crea automáticamente una copia.

## 7. Operaciones básicas con listas

Las listas son **secuencias**, por lo que admiten muchas de las mismas operaciones que las cadenas de texto (`str`).

- `+` concatena dos listas.
- `*` repite los elementos de una lista.
- `len()` devuelve el número de elementos.
- `in` comprueba si un elemento pertenece a una lista.
- `for` permite recorrer sus elementos.

**Importante:** Para concatenar con `+`, ambos operandos deben ser listas. No podemos sumar directamente una lista y una cadena de texto.


```python

# Longitud
print(len([1, 2, 3]))

# Concatenación
print([1, 2, 3] + [4, 5, 6])

# Repetición
print(['Ni!'] * 4)

# Pertenencia
print(3 in [1, 2, 3])

# Iteración
for x in [1, 2, 3]:
    print(x, end=' ')

print([1, 2] + list("34")) # Correcto
print([1, 2] + "34")       # TypeError    
```


## 8. Comprensión de listas (List Comprehensions)

Una **comprensión de listas** permite crear una nueva lista
aplicando una expresión a cada elemento de un iterable.

### Sintaxis

`[expresión for variable in iterable]`

- **expresión:** operación que se aplica a cada elemento.
- **variable:** representa cada elemento durante el recorrido.
- **iterable:** secuencia que recorremos (lista, cadena, range, etc.).

Es una alternativa más compacta al bucle `for` con `append()`.


```python

# Comprensión de listas
resultado = [c * 4 for c in 'SPAM']
print(resultado)

# Equivalente con un bucle for
resultado = []

for c in 'SPAM':
    resultado.append(c * 4)

print(resultado)
```


## 9. Función map()

`map()` aplica una función a cada elemento de un iterable.

**Sintaxis:**

`map(función, iterable)`

En Python 3, `map()` devuelve un iterador.
Podemos convertirlo en una lista mediante `list()`.


```python

numeros = [-1, -2, 0, 1, 2]

resultado = list(map(abs, numeros))

print(resultado)

```

¿Qué hace cada parte?
1. abs → Es la función que queremos aplicar.
2. map(abs, numeros) → Aplica abs() a cada elemento de numeros.
3. list() → Convierte el resultado de map() en una lista.
4. resultado = → Guarda esa lista en la variable.


## 10. Indexación y slicing

Las listas son secuencias ordenadas y permiten acceder a sus elementos mediante índices.

### Indexación

- `L[0]` → Primer elemento.
- `L[1]` → Segundo elemento.
- `L[-1]` → Último elemento.
- `L[-2]` → Penúltimo elemento.

**Los índices comienzan en 0.**

### Slicing (fragmentación)

Permite obtener una parte de la lista.

**Sintaxis:**

`L[inicio:fin:paso]`

- `inicio`: posición donde empieza.
- `fin`: posición donde termina (no incluida).
- `paso`: salto entre elementos.

El slicing devuelve una nueva lista.


```python
L = ['spam', 'Spam', 'SPAM']

# Indexación
print(L[2])     #
print(L[-2])    

# Slicing
print(L[1:])    
print(L[:2])    
print(L[::-1])  
```


## 11. Matrices y listas anidadas

Una **lista anidada** es una lista que contiene otras listas.

Podemos utilizar listas anidadas para representar matrices, tablas o cuadrículas.

### Acceso a elementos

`matriz[fila][columna]`

- El primer índice selecciona la fila.
- El segundo índice selecciona el elemento dentro de esa fila.

**Importante:** Tanto las filas como las columnas empiezan en el índice 0.


```python

matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matriz[1])     
print(matriz[1][1])  
print(matriz[2][0])  
```


## 12. Modificación de listas (Changing Lists in Place)

Las listas son **mutables**, es decir, podemos modificar sus elementos después de crearlas.

Estas modificaciones se realizan **in place** (sobre la propia lista), sin necesidad de crear una nueva.

### Modificación mediante índices

`L[i] = valor`

Sustituye el elemento que ocupa la posición `i`.

### Modificación mediante slicing

`L[inicio:fin] = [nuevos_elementos]`

Sustituye un fragmento de la lista por otros elementos.

**Importante:**
- Los índices empiezan en 0.
- El índice `fin` no se incluye.
- Podemos sustituir varios elementos a la vez.
- La cantidad de elementos nuevos puede ser distinta de la cantidad sustituida.


```python

L = ['spam', 'Spam', 'SPAM']

# Modificar un elemento por índice
L[1] = 'eggs'
print(L)
# ['spam', 'eggs', 'SPAM']

# Sustituir un fragmento de la lista
L[0:2] = ['eat', 'more']
print(L)
# ['eat', 'more', 'SPAM']
```


## 13. Asignación mediante slicing

El **slicing** permite modificar varios elementos de una lista mediante una asignación.

### Sintaxis

`L[inicio:fin] = iterable`

- `inicio`: posición donde comienza la modificación.
- `fin`: posición donde termina (no incluida).
- `iterable`: elementos que se incorporan a la lista.

### Operaciones importantes

| Operación | Descripción |
|---|---|
| `L[1:3] = [8, 9]` | Sustituye los elementos de los índices 1 y 2. |
| `L[1:1] = [8, 9]` | Inserta elementos en la posición 1 sin eliminar ninguno. |
| `L[:0] = [8, 9]` | Inserta elementos al principio. |
| `L[len(L):] = [8, 9]` | Añade elementos al final. |
| `L[1:3] = []` | Elimina los elementos de los índices 1 y 2. |

**Importante:** Cuando `inicio` y `fin` son iguales, el fragmento está vacío y podemos insertar elementos sin sustituir ninguno.


```python

L = [1, 2, 3]

L[:0] = [0]
print(L)


# Insertar en una posición
L[2:2] = ['nuevo']
print(L)


# Añadir al final
L[len(L):] = [4, 5]
print(L)


# Eliminar un fragmento
L[2:4] = []
print(L)

```


## 14. Métodos de las listas (List Method Calls)

Los **métodos** son funciones asociadas a un objeto.

Se llaman utilizando un punto:

`lista.metodo(argumentos)`

### Métodos principales

| Método | Descripción |
|---|---|
| `L.append(x)` | Añade un elemento al final de la lista. |
| `L.sort()` | Ordena los elementos de la lista. |
| `L.reverse()` | Invierte el orden de los elementos. |

### Características importantes

- `append()` añade **un único objeto**, aunque sea otra lista.
- `sort()` ordena la lista original.
- `reverse()` invierte la posición de sus elementos, pero no los ordena.
- Los tres métodos modifican la lista directamente (**in place**).
- Los tres devuelven `None`, no una lista nueva.

**Cuidado:** No debemos escribir `L = L.sort()` porque `L` pasaría a valer `None`.


```python

# APPEND
L = ['spam', 'eggs', 'ham']
L.append('please')
print(L)

# SORT
L = ['abc', 'ABD', 'aBe']
L.sort()
print(L)

# REVERSE
L = [1, 2, 3, 4]
L.reverse()
print(L)
```
