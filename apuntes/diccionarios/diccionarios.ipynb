# Diccionarios

## En un diccionario los elementos se recupera por el nombre(clave). No necesitas saber la posicion si conoces la clave.

Ejemplo lista:


```python
alumno = ["Carlos", 25, "Ourense"]

print(alumno[0])  #posicion 0
print(alumno[1])  #posicion 1
```

Ejemplo diccionario:

```python
alumno = {
    "nombre": "Carlos",
    "edad": 25,
    "ciudad": "Ourense"
}

print(alumno["nombre"])  
print(alumno["edad"])   
```

Es mas facil recordar el nombre/clave que la posicion. Cada dato tiene una etiqueta que lo representa:


```python
coche = {
    "marca": "Toyota",
    "modelo": "Corolla",
    "año": 2022,
    "precio": 18000
}
print(coche["marca"])  
print(coche["precio"])   
```

En una lista tendrias que recorrer los elementos uno a uno:

```python
alumnos = [
    {"dni": "123", "nombre": "Ana"},
    {"dni": "456", "nombre": "Luis"},
    {"dni": "789", "nombre": "Marta"}
]

for alumno in alumnos:
    if alumno["dni"] == "456":
        print(alumno["nombre"])
```

Con un diccionario esta tarea es mas sencilla:


```python
alumnos = {
    "123": "Ana",
    "456": "Luis",
    "789": "Marta"
}

print(alumnos["456"])
```

# Cambiando diccionarios in place:

Python busca la clave 'ham' y sustituye su valor 1 por una lista con tres elementos:

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
D['ham'] = ['grill', 'bake', 'fry']
D
```

Con 'del' se elimina la entrada del diccionario

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
del D['eggs']               
D
```

Se añade nueva entrada al diccionario:

```python
D['brunch'] = 'Bacon'
D
```

# Métodos de un objeto diccionario

## values 
Devuelve los valores del diccionario.

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
list(D.values())     
```

## items
Devuelve todas las tuplas (key,value)

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
list(D.items())
```

## get

Si buscamos un objeto que no existe nos dará un KeyError

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
D['toast']  
```

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
D.get('spam')
```

Si usamos get para buscarlo y el elemento no esta nos devolverá un None y no un error:

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
print(D.get('toast')) 
```

toast': es la clave que estás buscando.

88: es el valor que devolverá si esa clave no existe.

Como 'toast' no está en el diccionario, Python devuelve 88:

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}

D.get('toast', 88)          
```

Python recorre D2 y añade sus elementos a D.

'toast': 4 → No existía, así que lo añade.

'muffin': 5 → No existía, así que lo añade.

'ham': 10 → Ya existía, así que sustituye el valor anterior (1) por 10:

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
D2 = {'toast':4, 'muffin':5, 'ham': 10}
D.update(D2)
D
```

pop elimina el elemento del diccionario:

```python
D = {'spam': 2, 'ham': 10, 'eggs': 3, 'toast': 4, 'muffin': 5}
D.pop('muffin')

D.pop('toast')

D
```

Para cualquier diccionario D, **for key in** D funciona de la misma manera que for year in table()

```python
table = {'1975': 'Holy Grail', '1979': 'Life of Brian', '1983': 'The Meaning of Life'}
for year in table:                      
    print(year + '\t' + table[year])
```

```python
for key in table:
    print(key + '\t' + table[key])
```

```python
for key in table.keys():
    print(key + '\t' + table[key])
```
