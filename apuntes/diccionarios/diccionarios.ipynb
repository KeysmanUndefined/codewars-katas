# Diccionarios

## En un diccionario los elementos se recupera por el nombre(clave). No necesitas saber la posicion si conoces la clave.

Ejemplo lista:


```python
alumno = ["Carlos", 25, "Ourense"]

print(alumno[0])  #posicion 0
print(alumno[1])  #posicion 1
```

Ejemplo diccionario:

```python
alumno = {
    "nombre": "Carlos",
    "edad": 25,
    "ciudad": "Ourense"
}

print(alumno["nombre"])  
print(alumno["edad"])   
```

Es mas facil recordar el nombre/clave que la posicion. Cada dato tiene una etiqueta que lo representa:


```python
coche = {
    "marca": "Toyota",
    "modelo": "Corolla",
    "año": 2022,
    "precio": 18000
}
print(coche["marca"])  
print(coche["precio"])   
```

En una lista tendrias que recorrer los elementos uno a uno:

```python
alumnos = [
    {"dni": "123", "nombre": "Ana"},
    {"dni": "456", "nombre": "Luis"},
    {"dni": "789", "nombre": "Marta"}
]

for alumno in alumnos:
    if alumno["dni"] == "456":
        print(alumno["nombre"])
```

Con un diccionario esta tarea es mas sencilla:


```python
alumnos = {
    "123": "Ana",
    "456": "Luis",
    "789": "Marta"
}

print(alumnos["456"])
```

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}  
D['spam']
```

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
list(D.keys()) 
```

dict_keys([...])
Es un objeto iterable que representa las claves del diccionario.Respeta el orden en el que se insertaron las claves. Puedes extraer las claves:

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
print(D.keys())
```

# Cambiando diccionarios in place:

Python busca la clave 'ham' y sustituye su valor 1 por una lista con tres elementos:

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
D['ham'] = ['grill', 'bake', 'fry']
D
```

Con 'del' se elimina la entrada del diccionario

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
del D['eggs']               
D
```

Se añade nueva entrada al diccionario:

```python
D['brunch'] = 'Bacon'
D
```

# Métodos de un objeto diccionario

## values 
Devuelve los valores del diccionario.

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
list(D.values())     
```

## items
Devuelve todas las tuplas (key,value)

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
list(D.items())
```

## get

Si buscamos un objeto que no existe nos dará un KeyError

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
D['toast']  
```

```python
D = {'spam': 2, 'ham': 1, 'eggs': 3}
D.get('spam')
```

Si usamos get para buscarlo y el elemento no esta nos devolverá un None y no un error:

```python
print(D.get('toast')) 
```
