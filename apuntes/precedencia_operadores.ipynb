
## Orden de precedencia de operadores en Python

Los operadores se evalúan según su prioridad, de mayor a menor.

**1 = Mayor prioridad | 17 = Menor prioridad**

| Prioridad | Operadores | Descripción |
|---|---|---|
| 1 | `()` | Paréntesis y agrupación |
| 2 | `[]`, `.`, `()` | Indexación, atributos y llamadas a funciones |
| 3 | `**` | Potencia |
| 4 | `+x`, `-x`, `~x` | Signo positivo, negativo y negación de bits |
| 5 | `*`, `/`, `//`, `%`, `@` | Multiplicación, divisiones, módulo y multiplicación matricial |
| 6 | `+`, `-` | Suma y resta |
| 7 | `<<`, `>>` | Desplazamiento de bits |
| 8 | `&` | AND a nivel de bits |
| 9 | `^` | XOR a nivel de bits |
| 10 | `\|` | OR a nivel de bits |
| 11 | `==`, `!=`, `<`, `>`, `<=`, `>=`, `is`, `in` | Comparaciones y pertenencia |
| 12 | `not` | Negación lógica |
| 13 | `and` | AND lógico |
| 14 | `or` | OR lógico |
| 15 | `if ... else` | Expresión condicional |
| 16 | `lambda` | Función anónima |
| 17 | `:=` | Asignación mediante expresión |
