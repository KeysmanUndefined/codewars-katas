# Cap. 6 mutabilidad.


```python
a = 5  
b = a                 # b vale 5 porque a vale 5
print("b es ", b)

a = 3
b = a               # b vale 3 porque a vale 3. b vale el valor que tiene a antes de mutar a 'spam'
a = 'spam'          # b no vale 'spam'. 
print("b es ", b)
```

```python
a = 3
b = a
a = a + 2   
          # b sigue valiendo 3 porque a solo muta despues de que b coja el valor de a.
print("b es ", b)
```

```python
L = [2,3,4]
M = L
L[0] = 24  # Las listas son mutables. Esta instruccion muta la primera posicion de L y por lo tanto de M.

print("L = ", L) 
print("M = ", M)
```

```python
L = [2,3,4]
M = L[:]   #crea una copia de la lista
L[0] = 24   #modifica la lista original sin cambiar la copia

print("L = ", L)
print("M = ", M)
```

```python
L = [2,3,4]
M = L    #crea copia de la lista
L == M   #compara si un valor es igual al otro
L is M   #L es M? si, referencia al mismo valor
```

```python
L = [1, 2, 3]  #L referencia un objeto
M = [1, 2, 3]  #M referencia un objeto diferente a L aunque tenga el mismo valor

print(
    "L == M",L == M 

)

print(
    "L is M",L is M
)
```

```python
X = 42  #referencia un valor
Y = 42	#referencia mismo valor que X	    

X  == Y      # Compara los valores   
X is Y 		 # pregunta si x es y. Como sus valores son iguales es True  
```
