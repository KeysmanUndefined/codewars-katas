# **Polimorfismo**
  
El signo más + significa cosas diferentes para distintos objetos: suma para los números y concatenación para las cadenas de texto.

Esta es una propiedad general de Python llamada polimorfismo.

El significado de una operación depende de los objetos sobre los que se realiza.

Como veremos cuando estudiemos el tipado dinámico, esta propiedad de polimorfismo explica gran parte de la concisión y flexibilidad del código Python.

Como los tipos no están restringidos, una operación escrita en Python normalmente puede funcionar automáticamente con muchos tipos diferentes de objetos, siempre que estos admitan una interfaz compatible (como la operación + en este caso).

# **Inmutabilidad**

Los objetos inmutables no pueden modificarse.

Todos los objetos de Python se clasifican como inmutables (no modificables) o mutables (modificables).

En cuanto a los tipos principales:

los números, las cadenas y las tuplas son inmutables;

las listas, los diccionarios y los conjuntos son mutables.

```python
s[0] = "z"
```
