```python
x = 88 #esta es una variable global


def func():
    x = 99 #variable solo de la funcion

func()
print(x)  
```

```python
x = 88 			

def func():
	global x #hace referencia a la variable fuera de la funcion
	x = 99 	#modifica la vriable global dentro de la funcion	

func()
print("x =", x) 
```

```python
y,z = 1,2

def all_global():
    global x  #asigna una variable global
    x = y + z #la variable x global vale y + z (1+2)

all_global()
print('x = %d, y = %d, z = %d' % (x, y, z))
```

```python
x = 99 #Es la global.Esta fuera de la funcion

def f1():
    x = 88 #la local esta dentro de la funcion
    def f2():
        print("x local = ", x)
    f2()


f1()
print("x global = ", x) 
```

```python
def f1():
    x = 88 #variable local. Solo existe en la funcion
    def f2():
        print("x =", x) #imprime variable local
    return f2

action = f1()
action()
```

```python
def maker(n): #argumento n
    def action(x): #define action con argumento x
        return x ** n #devuelve x elevado a n
    return action devuelve action

f = maker(2) #n vale 2

f(3)  # pasa el argumento 3 y lo eleva a 2 = 9

f(4) #pasa el argumento 4 y en el return de la funcion lo devuelve elevado al argumento del maker()
```
