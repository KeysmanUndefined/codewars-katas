# Suma posible

## En este ejercicio se resuelve el ejercicio de codewars [Every possible sum of two digits - Codewars](https://www.codewars.com/kata/5b4e474305f04bea11000148)

## Solucion:

```python
def digits(n):
    
    n = str(n)
    resultado=[]
    
    for i in range(len(n)):
        
        for j in range(i + 1,len(n)):
            
            resultado.append(int(n[i]) + int(n[j]))
            
    
    return resultado
```

## Descripcion:

Dado un número >= 0, devuelve todas las posibles sumas de dos dígitos del mismo. En la salida, los pares de dígitos deben estar ordenados primero de izquierda a derecha para el primer dígito, y luego de izquierda a derecha para los dígitos subsiguientes. Por ejemplo, 12345: todas las posibles sumas de dos dígitos de ese número son: [ 1 + 2, 1 + 3, 1 + 4, 1 + 5, 2 + 3, 2 + 4, 2 + 5, 3 + 4, 3 + 5, 4 + 5 ]. Por lo tanto, el resultado debe ser: [ 3, 4, 5, 6, 5, 6, 7, 7, 8, 9 ].
