## 1. Paso de objetos inmutables:

```python
def f(a):
    a = 99
    print(a is b) # a no apunta al mismo objeto que b

b = 88		
f(b)			# a y b hacen referencica al mismo objeto en este caso 88	
print("b = ", b)

	
```

## 2. Paso de objetos mutables:

```python
def changer(a, b):	    
		a = 2			
		b[0] = 'spam' #sustituyo la posicion 0 de la lista.l=[1,2]. cambio posicion 0, el 1 lo cambio por 'spam' y queda definida la lista asi ['spam', 2]
		
			
x = 1
l = [1, 2]    #lista
changer(x, l)    #aplico la lista como argumento       
print("l = ", l)   #printo solo la lista      
assert x == 1, "x es 1"
```

## 3. Cómo evitar que una rutina modifique los elementos mutables:


### 3.1 Pasar una copia

```python
l = [1, 2]
changer(x, l[:]) # Pasa una copia de la lista a la función
                 # [:] crea una copia superficial de toda la lista

print("l = ", l) 
```

### 3.2 Utilizar invariantes (en las estructuras de datos de entrada)

```python
l = [1, 2]  #define lista entrada
def changer(a, b):
    b = b[:]	#crea copia de b
    a = 2
    b[0] = 'spam'	#modifica copia de b
    

changer(x, l)
assert l == [1, 2], "l es una invariante"
print("l = ", l) #es la misma lista de entrada. No la hemos modificado
```

### 3.3 Convertir el objeto mutable a un objeto inmutable:

```python
l = [1, 2]
def changer(a, b):	   
		a = 2			
		b[0] = 'spam'   # TypeError: las tuplas no permiten modificar sus elementos. Son inmutables


changer(x, tuple(l))	# Pasamos una tupla; al intentar modificarla, se produce un error

print("l = ", l)
assert l == [1, 2], "l es una invariante"
```

## 4. Keyword and Default Examples

### 4.1. Comportamiento por defecto:

```python
def f(a, b, c): 
    print(a, b, c)

f(1, 2, 3)      # paso argumento a la funcion
```

### 4.2. Keywords

```python
def f(a, b, c): 
    print(a, b, c)

f(c=3, b=2, a=1)    # 1 2 3
```

```python
f(1, c=3, b=2)		# se pasa el argumento por posicion
                    
```

### 4.3. Defaults

```python
def f(a, b=2, c=3): 	# necesita el argumento para a
	print(a, b, c)


f(1)        # 1 2 3
f(a=1)      # le pasa argumento a
```

```python
def f(a, b=2, c=3): 	# se requiere a como argumento
	print(a, b, c)
    
f(1, 4)     # le pasamos los argumentos a == 1 b == 4 y c es opcional pues ya tiene un valor
f(1, 4, 5)	# le pasamos los argumentos a == 1 b == 4 y c == 5

f(1, c=6)	# le pasamos los argumentos a y c
```

### 4.4 Combining keywords and defaults

```python
def func(spam, eggs, toast=0, ham=0):
	print((spam, eggs, toast, ham)) 	# se requieren los valores para spam y eggs. toast y ham son opcionales

func(1, 2)							# Output: (1, 2, 0, 0)
func(1, ham=1, eggs=0)				# Output: (1, 0, 0, 1)
func(spam=1, eggs=0)				# Output: (1, 0, 0, 0)
func(toast=1, eggs=2, spam=3)		# Output: (3, 2, 1, 0)
func(1, 2, 3, 4) 					# Output: (1, 2, 3, 4)
```

### 1.5. Python 3.X Keyword-Only Arguments

```python
def kwonly(a, *b, c): #Los parametros despues del * son obligatorios por nombre. 
    print(a, b, c)

kwonly(1, 2, c=3)       # paso parametro c3 por nombre

kwonly(a=1, c=3)        # paso parametro c3 por nombre

kwonly(1, 2, 3)         # TypeError: kwonly() missing 1 required keyword-only argument: 'c'. no lo paso por nombre
```

```python
def kwonly(a, *, b='spam', c='ham'): #b y c son opcionales porque tienen un valor. si quiero cambiar el valor se lo pasamos por nombre.
	print(a, b, c)

kwonly(1)               # 1 spam ham
kwonly(1, c=3)          # 1 spam 3
kwonly(a=1)             # 1 spam ham
kwonly(c=3, b=2, a=1)   # 1 2 3
kwonly(1, 2)            # TypeError: kwonly() takes 1 positional argument but 2 were given
```

```python
def kwonly(a, *, b, c='spam'):
    print(a, b, c)

kwonly(1, c='eggs')     # TypeError: solo le hemos pasado el argumento a y c, el b es obligatorio por nombre
```

## Las dos últimas extensiones relacionadas, * y **, están diseñadas para admitir funciones que reciben cualquier número de argumentos.

Ambas pueden aparecer tanto en la definición de una función como en una llamada a una función.

```python
def f(*args): # *args usa un numero indeterminado de argumentos y los almacena en una tupla
    print(args)

f()                 # ()
f(1,)               # (1,)
f(1, 2, 3, 4)       # (1, 2, 3, 4)  
```

```python
def f(**args): # **args recoge los argumentos en un diccionario en lugar de una tupla
    print(args)
f()                 # {}
f(a=1, b=2)         # {'a': 1, 'b': 2}
```

```python
def f(a, *pargs, **kargs): # *pargs recoge los argumentos restantes sin nombre y los almacena en un tupladiccionario
                           # y **kargs recoge los argumentos con nombre y los almacena en una diccionario
    print(a, pargs, kargs)
f(1, 2, 3, x=1, y=2)        # 1 (2, 3) {'y': 2, 'x': 1}
```

## Calls: Unpacking arguments

```python
def func(a, b, c, d): 
    print(a, b, c, d)

args = (1, 2)       #asigna argumentos (1, 2) a args
args += (3, 4)      #añade argumentos (3,4) a args
func(*args)         # *args aqui desempaqueta los argumentos para pasarlos a la funcion en orden
```

```python
def func(a, b, c, d): 
    print(a, b, c, d)
args = {'a': 1, 'b': 2, 'c': 3}
args['d'] = 4
func(**args)    # **args son los valores de la variable
func(*args)     # *args es el nombre de la variable
```

```python
func(*(1, 2), **{'d': 4, 'c': 3})       # Same as func(1, 2, d=4, c=3)
# 1 2 3 4
func(1, *(2, 3), **{'d': 4})            # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, c=3, *(2,), **{'d': 4})         # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
func(1, *(2, 3), d=4)                   # Same as func(1, 2, 3, d=4)
# 1 2 3 4
func(1, *(2,), c=3, **{'d':4})      # Same as func(1, 2, c=3, d=4)
# 1 2 3 4
```
