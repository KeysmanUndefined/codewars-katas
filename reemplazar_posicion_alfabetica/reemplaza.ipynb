# "Replace With Alphabet Position"

## En este ejercicio se resuelve el ejercicio de Coderwars "Replace With Alphabet Position"


[Ejercicio Alphabet Position](https://www.codewars.com/kata/546f922b54af40e1e90001da)

```python
def alphabet_position(text):
    
    resultado = ""
    text = text.lower()

    for i in text:
        
        if "a" <= i <= "z":
            
            resultado = resultado + str(ord(i) -96) + " "
            
    resultado = resultado.strip()
            
            
            
    return resultado
```
