[Ejercicio Permute a Palindrome - Codewars](https://www.codewars.com/kata/58ae6ae22c3aaafc58000079)

## Solucion:

```python
def permute_a_palindrome(input):
    
    impares = 0
    input = input.replace(" ", "")
    # set() crea un conjunto con las letras diferentes de la palabra, eliminando los duplicados.
    for letra in set(input):
        # .count() cuenta cada vez que aparece un elemento. Letra en este caso
        contador = input.count(letra)

        if contador % 2 != 0:
            impares += 1

    return impares <= 1
```

6	+, -	Suma y resta

8	&	AND a nivel de bits

11	==, !=, <, >, <=, >=, is, in	Comparaciones y pertenencia

13	and	AND lógico

15	if ... else	Expresión condicional
