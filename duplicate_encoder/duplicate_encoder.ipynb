#Solucion:


````python
def duplicate_encode(word):
    
    code = ""    
    word = word.lower()
    
    for letra in word:        
         if word.count(letra) > 1:           
            code = code + ")"
        
        else:            
            code = code + "("
            
    return code
````

## Descripcion:

El objetivo de este ejercicio es convertir una cadena de texto (`string`) en una nueva cadena en la que cada carácter se sustituye por:

- `"("` si el carácter aparece **una sola vez** en la cadena original.
- `")"` si el carácter aparece **más de una vez** en la cadena original.

Para determinar si un carácter está repetido, se deben **ignorar las diferencias entre mayúsculas y minúsculas**.

### Ejemplos

```python
"din"      =>  "((("
"recede"   =>  "()()()"
"Success"  =>  ")())())"
"(( @"     =>  "))(("
```

Si la condición se cumple, añadimos el carácter ")" a la variable code. Esto ocurre cada vez que el bucle encuentra una letra que aparece más de una vez en la cadena original.

````python
word = "onomatopoeia"
code = ""

for letra in word:
    if word.count(letra) > 1:
        code = code + ")"
        print(code)
````

code = code + ")" no sustituye el contenido anterior de code, sino que añade un nuevo paréntesis al final.


Primera vez: ")"
Segunda vez: "))"
Tercera vez: ")))"
Y así sucesivamente.

## 7. Else

Si la condición no se cumple, añadimos el carácter "(" a la variable code. Esto ocurre cada vez que el bucle encuentra una letra que no aparece más de una vez en la cadena original.

````python
word = "onomatopoeia"
code = ""

for letra in word:
    if word.count(letra) > 1:
        code = code + ")"

    else:
        code = code + "("

print(code)
````

Ahora tenemos 12 caracteres en la cadena original y 12 paréntesis en el resultado code. Cada carácter genera exactamente un paréntesis.
